# Zero-Shot Prompting with LangChain

This notebook demonstrates **Zero-Shot Prompting**, a fundamental prompt pattern where we ask the LLM to perform a task without providing any examples.

## Key Characteristics of Zero-Shot Prompting:
- No training examples provided
- Relies on the model's pre-trained knowledge
- Direct task instruction
- Minimal context setup

## Examples We'll Cover:
1. 🧮 Mathematical reasoning
2. 🏔️ Knowledge retrieval
3. 🏷️ Text classification
4. 📝 Creative writing
5. 🔄 Format conversion
6. 🎭 Role-based responses

```python
import os
from langchain.agents import AgentType, initialize_agent
from langchain_openai import ChatOpenAI
from langchain_community.tools import ShellTool
from langchain.schema import HumanMessage, SystemMessage
from langchain.prompts import PromptTemplate
from langchain.chains import LLMChain
import json
```

```python
shell_tool = ShellTool()
# Initialize the OpenAI LLM
# Make sure to set your OPENAI_API_KEY environment variable
api_key = os.getenv("OPENAI_API_KEY")
if not api_key:
    print("Warning: OPENAI_API_KEY environment variable not set")
    print("Please set it with: os.environ['OPENAI_API_KEY'] = 'your-api-key-here'")

llm = ChatOpenAI(
    temperature=0,
    openai_api_key=api_key,
    model="gpt-3.5-turbo",
    max_tokens=1000
)

print("LLM initialized successfully!")

shell_tool.description = shell_tool.description + f"args{shell_tool.args}".replace("{", "{{").replace("}", "}}")
```

```python
user_input = "What is 15 * 23 + 47?"

self_ask_with_search = initialize_agent([shell_tool], llm,
agent=AgentType.CHAT_ZERO_SHOT_REACT_DESCRIPTION, verbose=True)
self_ask_with_search.run(user_input)

# =============================================================================
# ZERO-SHOT PROMPTING DEMONSTRATION
# =============================================================================

print("🎯 ZERO-SHOT PROMPTING EXAMPLES")
print("="*50)

# Example 1: Mathematical Problem Solving
print("\n📊 Example 1: Mathematical Reasoning")
print("-" * 30)

math_prompt = """Solve this step by step: What is 15 × 23 + 47?

Please show your work and provide the final answer."""

response = llm.invoke([HumanMessage(content=math_prompt)])
print(f"Question: {math_prompt}")
print(f"\nAnswer: {response.content}")
```

```python
# Let's try a direct knowledge question instead of web scraping
user_input_mountains = "Name the tallest mountain in the world and in North America"

# Using the agent without shell tools to avoid large web content
from langchain.agents import create_chat_agent, AgentExecutor
from langchain import hub

# Create a simple agent without shell tools
simple_llm = ChatOpenAI(temperature=0, openai_api_key=os.getenv("OPENAI_API_KEY"), model="gpt-3.5-turbo")
response = simple_llm.invoke(user_input_mountains)
print(response.content)

# Example 2: Knowledge Retrieval
print("\n🏔️ Example 2: Knowledge Retrieval")
print("-" * 30)

knowledge_prompt = """Name the tallest mountain in the world and the tallest mountain in North America. 
Provide their heights and a brief fact about each."""

response = llm.invoke([HumanMessage(content=knowledge_prompt)])
print(f"Question: {knowledge_prompt}")
print(f"\nAnswer: {response.content}")

# Example 3: Classification Task
print("\n🏷️ Example 3: Text Classification")
print("-" * 30)

classification_prompt = """Classify the following text as Positive, Negative, or Neutral sentiment:

"I absolutely love this new coffee shop! The atmosphere is cozy and the baristas are incredibly friendly. The coffee tastes amazing too!"

Sentiment:"""

response = llm.invoke([HumanMessage(content=classification_prompt)])
print(f"Text to classify: \"I absolutely love this new coffee shop! The atmosphere is cozy and the baristas are incredibly friendly. The coffee tastes amazing too!\"")
print(f"\nClassification: {response.content}")
```

```python
# Example 4: Creative Writing
print("\n✍️ Example 4: Creative Writing")
print("-" * 30)

creative_prompt = """Write a short haiku about artificial intelligence."""

response = llm.invoke([HumanMessage(content=creative_prompt)])
print(f"Request: {creative_prompt}")
print(f"\nHaiku: {response.content}")

# Example 5: Format Conversion
print("\n🔄 Example 5: Format Conversion")
print("-" * 30)

format_prompt = """Convert the following information into JSON format:

Name: Sarah Johnson
Age: 32
Occupation: Software Engineer
City: Seattle
Hobbies: Photography, Hiking, Reading

JSON:"""

response = llm.invoke([HumanMessage(content=format_prompt)])
print(f"Original data: Name: Sarah Johnson, Age: 32, Occupation: Software Engineer, City: Seattle, Hobbies: Photography, Hiking, Reading")
print(f"\nJSON format: {response.content}")
```

```python
# Example 6: Role-Based Zero-Shot Prompting
print("\n🎭 Example 6: Role-Based Response")
print("-" * 30)

role_prompt = """You are a friendly fitness coach. Someone asks: "I want to start exercising but I'm a complete beginner and feel overwhelmed. What should I do?"

Provide encouraging advice in a supportive tone."""

response = llm.invoke([HumanMessage(content=role_prompt)])
print(f"Role: Fitness Coach")
print(f"Question: I want to start exercising but I'm a complete beginner and feel overwhelmed. What should I do?")
print(f"\nAdvice: {response.content}")
```

```python
# =============================================================================
# USING LANGCHAIN PROMPT TEMPLATES FOR ZERO-SHOT PROMPTING
# =============================================================================

print("\n\n🛠️ USING PROMPT TEMPLATES")
print("="*50)

# Create a reusable zero-shot classification template
classification_template = PromptTemplate(
    input_variables=["text", "categories"],
    template="""Classify the following text into one of these categories: {categories}

Text: {text}

Category:"""
)

# Create an LLM Chain
classification_chain = LLMChain(llm=llm, prompt=classification_template)

# Test with different inputs
test_cases = [
    {
        "text": "The movie was absolutely terrible. I couldn't wait for it to end.",
        "categories": "Positive, Negative, Neutral"
    },
    {
        "text": "Python is a versatile programming language used for web development, data science, and automation.",
        "categories": "Technology, Sports, Entertainment, Politics"
    }
]

print("\n📝 Template-based Classification Examples:")
for i, case in enumerate(test_cases, 1):
    result = classification_chain.run(text=case["text"], categories=case["categories"])
    print(f"\nExample {i}:")
    print(f"Text: {case['text']}")
    print(f"Categories: {case['categories']}")
    print(f"Classification: {result.strip()}")
```

## 🔍 Zero-Shot vs Few-Shot Prompting

### Zero-Shot Prompting (What we just demonstrated):
- ❌ **No examples provided**
- ✅ Direct task instruction
- ✅ Relies on model's pre-trained knowledge
- ✅ Quick and simple
- ⚠️ May be less accurate for complex tasks

### Few-Shot Prompting (Alternative approach):
- ✅ **Examples provided** in the prompt
- ✅ Model learns pattern from examples
- ✅ Often more accurate
- ⚠️ Requires more tokens/context

## 🎯 When to Use Zero-Shot Prompting:
1. **Simple, well-defined tasks** (math, basic classification)
2. **Common knowledge questions**
3. **When you want to save tokens/context**
4. **Quick prototyping and testing**
5. **Tasks the model was explicitly trained for**

## 💡 Tips for Effective Zero-Shot Prompting:
1. **Be specific and clear** in your instructions
2. **Use role-based prompts** when appropriate
3. **Specify the desired output format**
4. **Include context** that helps the model understand the task
5. **Test with different phrasings** if results aren't satisfactory
